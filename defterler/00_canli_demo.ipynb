# Canlı Demo — Baret Var mı? Başta mı?

Bir İSG kamerasının **"baretsiz çalışan"** uyarısı verebilmesi için iki karar gerekir:
1. **Görüntüde baret var mı?** → RGB / HSV renk maskesi (renkli alan %)
2. **Baret başta mı?** → maskenin elle işaretlenmiş baş bölgesiyle örtüşmesi → dört sınıflı karar

Renk maskesi yalnızca *renkli bölgeyi* bulur, *baret nesnesini* tanımaz. Ayarlar G1/G2'de seçilip dondurulmuştur; demo sırasında değiştirilmez.
