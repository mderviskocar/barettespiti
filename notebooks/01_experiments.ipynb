# Baret Tespiti ve Sınıflandırması

Bu dosya yalnızca deney akışı şablonudur. Henüz deney kodu, veri veya çıktı içermez. Uygulama sırasında ortak işlevler `src/baret_tespiti/` altına yazılacaktır. Deneyleri çalıştırmadan önce `docs/experiment_protocol.md` okunmalıdır.

## 1. Ortam, veri ve ayarlar
Görüntü kaynakları, 2 geliştirme / 4 kontrol ayrımı, doğru sınıflar ve elle işaretlenen baş bölgeleri burada gösterilecek.

## 2. Deney öncesi tahminler
Her deneyin iki cümlelik tahmini çalıştırmadan önce `docs/predictions.csv` dosyasına kaydedilecek.

## 3. Geliştirme: RGB/HSV eşikleri ve sınıf kuralı
Yalnızca geliştirme verisi kullanılacak. Seçilen ayarlar `configs/experiment.json` içinde sabitlenecek.

## 4. Kontrol: sekiz maske ve sınıf kararları
Dört kontrol görüntüsünde RGB/HSV maskeleri, alan oranları ve doğru/yanlış sınıf kararları gösterilecek.

## 5. Tek değişkenli deneyler
Bulanıklık ve küçültme ayrı ayrı uygulanacak; ayarlar ve değişmeyen koşullar kaydedilecek.

## 6. Hazır dedektör karşılaştırması (isteğe bağlı)
Kullanılırsa model kaynağı, sınıflar, giriş boyutu, kutular ve IoU >= 0.50 ile TP/FP/FN sayımları gösterilecek.

## 7. Hata analizi ve sonuçların dışa aktarılması
Yanlış alarm, kaçırma, nedenleri, sonraki kontroller ve sınırlılıklar açıklanacak. Ölçülmüş sonuçlar `outputs/results.csv`, görseller `outputs/figures/` altına kaydedilecek.
